```python
import pandas as pd
```

```python
tr_data = pd.read_csv("clean_data/temiz_turkce_data.csv")
en_data = pd.read_csv("clean_data/temiz_ingilizce_data.csv")
```

```python
tr_data.head()
```

```python
en_data.head()
```

```python
en_data["label"].value_counts()
```

```python
tr_positive = tr_data[tr_data["label"] == 1].sample(n=50000, random_state=42)
tr_negative = tr_data[tr_data["label"] == 0].sample(n=50000, random_state=42)
```

```python
tr_positive
```

```python
tr_negative
```

```python
tr_final = pd.concat([tr_positive,tr_negative])
tr_final.head()
```

```python
tr_final = tr_final.sample(frac=1, random_state=42).reset_index(drop=True)

tr_final.head()
```

```python
tr_final["label"].value_counts()
```

```python
final_data = pd.concat([tr_final, en_data],ignore_index=True)
final_data = final_data.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
final_data.head()
```

```python
print("Sınıf Dağılımı:\n", final_data['label'].value_counts())
print("\nToplam Veri Sayısı:", len(final_data))
```

```python
final_data.rename(columns={'clean_text': 'description'}, inplace=True)

final_data.head()
```

```python
nan_rows = final_data[final_data.isnull().any(axis=1)]

print(nan_rows)
print(nan_rows.index)
final_data = final_data.dropna(subset=["description"])
```

```python
final_data.to_csv("clean_data/temiz_genel_veri.csv", index=False)
print("Birleşik Veri Kaydedildi")
```
