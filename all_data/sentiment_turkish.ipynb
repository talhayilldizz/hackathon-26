```python
def clean_text(text):
    if not isinstance(text, str):
        return ""
    
    text = text.replace('İ', 'i').replace('I','ı').lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = text.strip()

    return text
```

```python
sentiment_preserving_words = {
    'değil', 'yok', 'hayır', 'ama', 'fakat', 'ancak', 'lakin', 'asla', 'hiç', 
    'en', 'çok', 'daha', 'pek', 'gayet', 'oldukça', 'aşırı', 'az', 'fazla', 
    'yeterli', 'eksik', 'maalesef', 'zira', 'çünkü', 'rağmen', 'belki'
}

nltk.download('stopwords')
tr_stopwords = set(stopwords.words("turkish"))
```

```python
tr_stopwords
```

```python
len(tr_stopwords)
```

```python
tr_stopwords = tr_stopwords - sentiment_preserving_words
len(tr_stopwords)
```

```python
def stopwords_clear(text):
        words = text.split()
        clean_words = [w for w in words if w not in tr_stopwords]
        return " ".join(clean_words)
```

```python
data['clean_text'] = data['combined_text'].apply(lambda x: stopwords_clear(clean_text(x)))
```
